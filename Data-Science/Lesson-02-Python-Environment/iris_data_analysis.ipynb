```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

df = pd.read_csv(url, names=columns)

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()
```

```python
print("Statistical Summary:")
print(df.describe())
```

```python
print("Flower Species Count:")
print(df["species"].value_counts())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
average_petal_length = np.mean(df["petal_length"])

print("Average Petal Length:",
      round(average_petal_length, 2), "cm")
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="species")

plt.title("Distribution of Iris Flower Species")
plt.xlabel("Flower Species")
plt.ylabel("Number of Flowers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="species", y="petal_length")

plt.title("Petal Length Comparison by Species")
plt.xlabel("Flower Species")
plt.ylabel("Petal Length (cm)")

plt.show()
```
