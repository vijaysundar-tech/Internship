```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"

df = pd.read_csv(url, sep=";")

print("First 5 rows of Red Wine Dataset:")
display(df.head())

print("\nDataset Shape:", df.shape)

print("\nWine Quality Distribution:")
print(df["quality"].value_counts().sort_index())
```

```python
plt.figure(figsize=(8, 5))

df.groupby("quality")["alcohol"].mean().plot(
    kind="bar",
    title="Average Alcohol Content by Wine Quality"
)

plt.xlabel("Wine Quality Score")
plt.ylabel("Average Alcohol Content (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="quality")

plt.title("Distribution of Wine Quality Scores")
plt.xlabel("Wine Quality")
plt.ylabel("Number of Wine Samples")

plt.tight_layout()
plt.show()
```

## Findings and Observations

1. The Red Wine Quality dataset contains 1,599 records and 12 columns.
2. Most wine samples have quality scores of 5 and 6.
3. Quality score 5 has the highest number of samples (681).
4. Quality score 8 has the highest average alcohol content among the quality groups.
5. The charts show that average alcohol content generally increases with wine quality score.
