```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python

import urllib.request
import zipfile
import pandas as pd

url = "https://archive.ics.uci.edu/static/public/320/student+performance.zip"

urllib.request.urlretrieve(url, "student.zip")

with zipfile.ZipFile("student.zip", "r") as zip_ref:
    zip_ref.extractall("student_data")

print("Dataset downloaded and extracted successfully!")
```

```python
import os

print("Current folder:", os.getcwd())

print("\nFiles in current folder:")
print(os.listdir())

print("\nFiles in student_data:")
if os.path.exists("student_data"):
    print(os.listdir("student_data"))
else:
    print("student_data folder not found")
```

```python
import zipfile
import os

with zipfile.ZipFile("student.zip", "r") as zip_ref:
    zip_ref.extractall("student_data")

print("Extraction completed!")

print("\nExtracted files:")
for root, dirs, files in os.walk("student_data"):
    for file in files:
        print(os.path.join(root, file))
```

```python
import zipfile
import os

inner_zip = "student_data/student.zip"

with zipfile.ZipFile(inner_zip, "r") as zip_ref:
    zip_ref.extractall("student_data")

print("Inner ZIP extracted successfully!")

for root, dirs, files in os.walk("student_data"):
    for file in files:
        print(os.path.join(root, file))
```

```python
df_csv = pd.read_csv(
    "student_data/student-mat.csv",
    sep=";"
)

df_csv.to_excel("student_performance.xlsx", index=False)

df = pd.read_excel("student_performance.xlsx")

print("Excel dataset imported successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nDataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python

marks = df["G3"].to_numpy()

print("Mathematics Performance Analysis")
print("Average Marks:", round(np.mean(marks), 2))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Median Marks:", np.median(marks))
```

```python
subjects = ["First Period", "Second Period", "Final"]

averages = [
    df["G1"].mean(),
    df["G2"].mean(),
    df["G3"].mean()
]

plt.figure(figsize=(8, 5))
plt.bar(subjects, averages)

plt.title("Average Student Marks")
plt.xlabel("Examination")
plt.ylabel("Average Marks")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Final Mathematics Marks")
plt.xlabel("Final Marks")
plt.ylabel("Number of Students")

plt.show()
```

## Findings and Observations
